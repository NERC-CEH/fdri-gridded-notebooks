## What this notebook does

This notebook explores **HadUK-Grid Rainfall**. [HadUK-grid](https://dx.doi.org/10.5285/f02cc6ddd92f45b18b9ab6ab544df7d9) is an observations-derived gridded dataset for the UK produced by the [UK Met Office](https://www.metoffice.gov.uk/research/climate/maps-and-data/data/haduk-grid/haduk-grid). It consists of many meteorological variables at various temporal and spatial resolutions. Here we will explore the daily, 1km rainfall dataset which runs from 1891.

Simple examples are shown for exploring and working with the dataset, which can be generalised to any gridded dataset stored in NetCDF format.

A key limitation of working with NetCDF is that data files have to be downloaded to your machine - or wherever you are running the notebook - before you can work with them. This is why we also provide a [version](https://colab.research.google.com/github/NERC-CEH/fdri-gridded-notebooks/blob/main/notebooks/HadUK-Grid/hadukgrid_zarr_R.ipynb) of this notebook that uses the Zarr format instead, which removes the limitation and allows you to work with multi-terabyte datasets without worrying about having to download it all.

**We will:**
1. Show how to open a dataset stored in NetCDF format
2. Show how to extract and plot a time series at a single location, and demonstrate the limitations of NetCDF for this
3. Show how to plot a map of rainfall at a single time step


**Data store:**
The data is currently publicly available via the CEDA Archive: https://dx.doi.org/10.5285/f02cc6ddd92f45b18b9ab6ab544df7d9

## 1. Download the NetCDF file

To work with NetCDF data we have to download the files ourselves, to our local machines (desktops/laptops), or wherever this notebook is running. A dataset in NetCDF format is usually split up into several files, typically across the time dimension.

To see what files are available, first go to the [catalogue web-page](https://dx.doi.org/10.5285/f02cc6ddd92f45b18b9ab6ab544df7d9) for this dataset. You will have to create a CEDA account and log in with it before you can see the files. Once done, a blue 'Download' button will appear underneath the dataset title. From there, navigate through the folder structure via 'rainfall', 'day', then 'latest', to see all the gridded rainfall files.

Looking at the filenames tells you that the dataset is split into monthly files across the time span it covers. Looking at the sizes in the 2nd column tells you that each file is at least 50MB. You can also see that there are a lot of files.

We will therefore download only a single file of the data. To do this in R you first need to obtain a token (the equivalent of logging-in) from the CEDA website using the below code snippet. You will need to replace "USERNAME" and "PASSWORD" with your actual username and password, keeping the quotes. Due to a limitation with the CEDA Server that manages tokens, your password will need to not contain any special characters, i.e. just be letters and numbers.

**IMPORTANT NOTE:** DO NOT SAVE THIS NOTEBOOK ANYWHERE PUBLIC WITH YOUR ACTUAL USERNAME AND PASSWORD WRITTEN IN. Your personal google drive should be fine, but it is still recommended you remove your actual username and password from the notebook before saving it.


```r
library(httr)
library(jsonlite)

url <- "https://services.ceda.ac.uk/api/token/create/"

username <- "USERNAME"
password <- "PASSWORD"
logindetails <- base64enc::base64encode(charToRaw(paste0(username, ":", password)))

response <- POST(
  url,
  add_headers(Authorization = paste("Basic", logindetails))
)

response_text <- content(response, as = "text", encoding = "UTF-8")
print(response_text)

if (status_code(response) == 200) {
  print('Token generated')
  response_data <- fromJSON(response_text)
  token <- response_data$access_token
} else {
  print('Token not generated')
}
```

Now you can download one of the files from the datasets by right clicking on the file of interest (in this case for Jan 2000) and selecting 'Copy Link'. This link is used below to download the file.

> **Note:** The file is ~60 MB. The download will likely take a few seconds!

```r
url <- "https://dap.ceda.ac.uk/badc/ukmo-hadobs/data/insitu/MOHC/HadOBS/HadUK-Grid/v1.3.1.ceda/1km/rainfall/day/v20250415/rainfall_hadukgrid_uk_1km_day_20000101-20000131.nc?download=1"

response <- GET(
  url,
  add_headers(Authorization = paste("Bearer", token))
)

nc_filename <- "rainfall_hadukgrid_uk_1km_day_20000101-20000131.nc"
writeBin(content(response, as = "raw"), nc_filename)
```

## 2. Open the file and inspect metadata

```r
nc <- nc_open(nc_filename)

# Print the full file summary — similar to ncdump -h
cat("=== FILE SUMMARY ===\n")
cat("Format   :", nc$format, "\n")
cat("Variables:", nc$nvars, "\n")
cat("Dims     :", nc$ndims, "\n")
```

### 2a. Global attributes (dataset-wide metadata)

```r
cat("=== GLOBAL ATTRIBUTES ===\n")
gatts <- ncatt_get(nc, 0)
for (nm in names(gatts)) {
  val <- as.character(gatts[[nm]])
  cat(sprintf("  %-30s : %s\n", nm, val))
}
```

### 2b. Variables

```r
cat("=== VARIABLES ===\n")
for (v in nc$var) {
  dim_str <- paste(sapply(v$dim, function(d) d$name), collapse = " × ")
  units   <- ncatt_get(nc, v$name, "units")$value
  lname   <- ncatt_get(nc, v$name, "long_name")$value
  cat(sprintf("  %-30s [%s]\n", v$name, dim_str))
  if (!is.logical(units)) cat(sprintf("    units     : %s\n", units))
  if (!is.logical(lname)) cat(sprintf("    long_name : %s\n", lname))
}
```

### 2c. Dimensions

```r
cat("=== DIMENSIONS ===\n")
for (dim in nc$dim) {
  cat(sprintf("  %-10s : %d values\n", dim$name, dim$len))
}
```

### 2d Read coordinate arrays

```r
x    <- ncvar_get(nc, "projection_x_coordinate")     # easting, metres
y    <- ncvar_get(nc, "projection_y_coordinate")     # northing, metres
time <- ncvar_get(nc, "time")  # time values

# Decode time axis
time_units <- ncatt_get(nc, "time", "units")$value

# Parse "hours since YYYY-MM-DD" format
origin_str <- sub(".*since\\s+", "", time_units)
origin_dt  <- as.POSIXct(origin_str, tz = "UTC",
                format = ifelse(grepl(":", origin_str), "%Y-%m-%d %H:%M:%S", "%Y-%m-%d"))

# Determine units multiplier (hours → seconds; days → seconds; etc.)
unit_word  <- tolower(sub("\\s+since.*", "", time_units))
multiplier <- switch(unit_word,
  "hours"   = 3600,
  "days"    = 86400,
  "minutes" = 60,
  "seconds" = 1,
  3600  # default to hours
)

datetimes <- origin_dt + as.numeric(time) * multiplier

cat(sprintf("x (easting) : %d values, %.0f m to %.0f m  (%.0f m spacing)\n",
  length(x), min(x), max(x), x[2] - x[1]))
cat(sprintf("y (northing): %d values, %.0f m to %.0f m  (%.0f m spacing)\n",
  length(y), min(y), max(y), y[1] - y[2]))
cat(sprintf("Time        : %d steps, %s to %s\n",
  length(datetimes), min(datetimes), max(datetimes)))
```

## 3. Time series at a single location

We find the nearest grid cell to a target x/y coordinate, then read all the timesteps for this file.

**Key Point:** To plot different months we would have to download the relevant files of the dataset as we did in Section 1. If we wanted to plot multiple months we would have to download multiple large files, if not the whole dataset, which quickly gets unwieldy given the size of the dataset. For increased flexibility and ease-of-use, consider using the Zarr version of the dataset instead via [this notebook](https://colab.research.google.com/github/NERC-CEH/fdri-gridded-notebooks/blob/main/notebooks/HadUK-Grid/hadukgrid_zarr_R.ipynb).

Load the data:

```r
# Target location in British National Grid coordinates (easting, northing in metres)
# Edinburgh ~ (325000, 673000)
target_x <- 325000
target_y <- 673000

# Find the nearest grid cell indices
i_x <- which.min(abs(x - target_x))
i_y <- which.min(abs(y - target_y))

cat(sprintf("Target  : x=%d, y=%d\n", target_x, target_y))
cat(sprintf("Nearest : x=%.0f (index %d), y=%.0f (index %d)\n",
  x[i_x], i_x, y[i_y], i_y))

# Read all time steps for this single grid cell
# start = [x_index, y_index, time_index_1], count = [1, 1, all_times]
ts_vals <- ncvar_get(
  nc, "rainfall",
  start = c(i_x, i_y, 1),
  count = c(1, 1, length(time))
)

ts_df <- data.frame(date = datetimes, rainfall_mm = as.numeric(ts_vals))
print(ts_df)
```

```r
units    <- ncatt_get(nc, "rainfall", "units")$value
longname <- ncatt_get(nc, "rainfall", "long_name")$value

barplot(
  ts_df$rainfall_mm,
  names.arg = format(ts_df$date, "%d"),
  col  = "blue",
  main = paste("HadUK-Grid Rainfall"),
  xlab = format(datetimes[1], "%b-%Y"),
  ylab = paste0(longname, " (", units, ")"),
)
```

## 5. Map at a single time step

Read the full 2-D spatial grid for the first time step (Jan 1st 2000) and plot it.

```r
# Read spatial slice for the first time step
# start = [all_x, all_y, time_step_1], count = [all, all, 1]
t_index <- 1
grid <- ncvar_get(
  nc, "rainfall",
  start = c(1, 1, t_index),
  count = c(-1, -1, 1)        # -1 means read all
)

cat(sprintf("Grid dimensions: %d x (easting) × %d y (northing)\n", dim(grid)[1], dim(grid)[2]))
```

```r
grid_plot <- grid

# Colour palette: white → blue
rain_cols <- colorRampPalette(c("white", "#c6dbef", "#4393c3", "#08306b"))(100)

image(
  x = x / 1000,     # convert metres to km for cleaner axis labels
  y = y / 1000,
  z = grid_plot,
  col  = rain_cols,
  xlab = "Easting (km)",
  ylab = "Northing (km)",
  main = paste0("HadUK-Grid Rainfall", "\n", format(datetimes[t_index], "%d %B %Y"))
)

# Simple colourbar using legend
legend("bottomright",
  legend = round(seq(min(grid, na.rm=TRUE), max(grid, na.rm=TRUE), length.out = 5)),
  fill   = rain_cols[c(1, 25, 50, 75, 100)],
  title  = units, bty = "n", cex = 0.8
)
```

```r
# ── Build long-format data frame for ggplot2 ──────────────────────────────────
map_df <- expand.grid(x = x, y = y)
map_df$value <- as.vector(grid)

# Get GB/Ireland coastline for overlay
gb <- ne_countries(country = c("United Kingdom", "Ireland"),
                   scale = "medium", returnclass = "sf")

# Clip to the dataset's geographic extent
x_range <- range(x)
y_range <- range(y)

p_map <- ggplot() +
  geom_raster(data = map_df, # specify the data to plot, and it's axes
              aes(x = x, y = y, fill = value), na.rm = TRUE) +
  scale_fill_distiller( # customise how the data is displayed
    palette   = "YlGnBu",
    direction = 1,
    name      = units,
    na.value  = NA,
    labels    = scales::comma
  ) +
  coord_sf( # coordinate display customisation
    expand = FALSE
  ) +
  labs( # label customisation
    title    = paste("HadUK-Grid Daily Rainfall"),
    subtitle = format(datetimes[t_index], "%d %B %Y"),
    x = "x", y = "y",
    caption  = "Source: FDRI / UKCEH · fdri-o.s3-ext.jc.rl.ac.uk"
  ) +
  theme_minimal(base_size = 12) # overall theme customisation


print(p_map)
```
